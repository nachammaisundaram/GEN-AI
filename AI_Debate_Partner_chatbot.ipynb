```python
import google.generativeai as genai
from google.generativeai import types
from google.colab import userdata

!pip install -q -U google-generativeai
```

```python
client=genai.Client(api_key=userdata.get('Ragproject'))
```

```python
prompt="""Act as an experienced public speaker and debate partner.
sponsibilities:
* Understand the debate topic and always take the opposite side of the user's opinion.
* Give short, formal, logical, and realistic arguments.
* Do not agree with the user's arguments.
* Do not repeat points already mentioned.
* Stay relevant to the current debate and remember the conversation context.
* Do not invent facts, statistics, or examples.
* If the user says "exit", end the debate politely."""

chat=client.chats.create(
    model='gemini-3.5-flash-lite',
    config=types.GenerateContentConfig(
        system_instruction=prompt,
    ),
    history=[]
)

print("=== AI Debate Partner ===")
print("Bot is ready for a debate!")
print("Type 'exit' or 'bye' to end the conversation.\n")

title=input("Enter a debate topic:")

initial_response = chat.send_message(title)
print(f"Bot:{initial_response.text}")

while True:
  user_input=input("User:")
  if user_input.lower() in['exit','bye']:
    print("Bot:Thank you for the debate! Hope you found it interesting. Have a nice day!")
    break

response = chat.send_message(user_input)
print(f"Bot:{response.text}")
```
