```python
!pip install -q google-genai faiss-cpu

import faiss
import numpy as np

from google import genai
from google.genai import types
from google.colab import userdata

client=genai.Client(api_key=userdata.get('Ragproject'))

EMBEDDING_MODEL='gemini-embedding-001'
EMBEDDING_DIMENSION=768
```

```python
def embed(text):
  response=client.models.embed_content(
      model=EMBEDDING_MODEL,
      contents=text,
      config=types.EmbedContentConfig(
          output_dimensionality=EMBEDDING_DIMENSION
      )
  )
  return(response.embeddings[0].values)
```

```python
description=[
    'Build REST APIs using Python, FastAPI, PostgreSQL and Docker.',
    'Develop machine learning models using Python, scikit-learn, pandas and NumPy',
    'Build LLM applications using Python, embeddings, RAG and vector databases.',
    'Deploy and manage applications using AWS, Docker, Kubernetes and cloud infrastructure.',
    'Automate CI/CD pipelines using Jenkins, Docker, Git and Kubernetes.',
    'Analyze datasets using Python, SQL, Excel and Power BI to generate business insights.',
    'Build responsive web applications using React, JavaScript, HTML and CSS.',
    'Develop web applications using React, Node.js, Express and MongoDB.',
    'Develop AI-powered applications using Python, APIs, machine learning and cloud services.',
    'Design and optimize databases using SQL, PostgreSQL, database queries and data modeling.'
]

jobs = [
    "Python Backend Developer",
    "Machine Learning Engineer",
    "Generative AI Engineer ",
    "Cloud Engineer",
    "DevOps Engineer",
    "Data Analyst",
    "Frontend Developer",
    "Full Stack Developer",
    "AI Software Engineer",
    "Database Developer"
]

des_vector = np.array([embed(des) for des in description])
print("Description vector shape: ",des_vector.shape)

des_vector = des_vector/np.linalg.norm(des_vector, axis=1, keepdims=True)
print("Normalized Description vector shape: ",des_vector.shape)


def search(prompt,top=1):
  query_vector=embed(prompt)
  query_vector=query_vector/np.linalg.norm(query_vector)

  score = des_vector @ query_vector

  top_score=np.argsort(score)[::-1][:top]
  return[
      {
          'job':jobs[i],
          'description':description[i],
          'score':float(score[i])
      }
      for i in top_score
  ]


#TEST
user_input=[
      'I am a fresher with Python programming skills and experience building Generative AI applications using embeddings and APIs.',
      'I have experience with AWS, Docker, Kubernetes and deploying applications through cloud infrastructure.',
      'I know Python, SQL, Excel and Power BI and I enjoy analyzing datasets and creating business insights.',
      'I have experience developing web applications using React, JavaScript, HTML and CSS.',
      'I can develop backend APIs using Python and FastAPI, work with databases and deploy applications using Docker.',
      'I am interested in fashion design, photography, illustration and creating digital artwork.'
  ]

for i, msg in enumerate(user_input, 1):
  print(f"{i}. {msg}")

  for output in search(msg):
    print(f"score:{output['score']:.4f} - {output['job']}\n")
```
