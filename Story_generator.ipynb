```python
client=genai.Client(api_key=userdata.get('Ragproject'))
```

```python
def story_generator(title, genre, length):

    prompt = f"""Generate a short story based on the
    title={title}, genre={genre}, and length={length}.
    Use simple English."""

    response = client.models.generate_content(
        model='gemini-3.5-flash-lite',
        contents=prompt,
        config=types.GenerateContentConfig(
            temperature=0.5,
            max_output_tokens=length,
        )
    )

    return response.text


title = input("Enter a title: ")
genre = input("Enter a genre: ")
length = int(input("Enter length (in tokens, e.g. 200): "))

story = story_generator(title, genre, length)

print(story)
```
