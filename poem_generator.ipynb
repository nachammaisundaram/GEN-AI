```python
from google import genai
from google.genai import types
from google.colab import userdata

client=genai.Client(api_key=userdata.get('Ragproject'))
```

```python
def poem_generator(title,genre,length):
  prompt=f"""generate a poem based on the title={title}, genre={genre}, length={length}.
             use simple english"""

  MODEL='gemini-3.5-flash-lite'

  response=client.models.generate_content(
      model=MODEL,
      contents=prompt,
      config=types.GenerateContentConfig(
          temperature=0.5,
          max_output_tokens=200,
      )
  )
  return response.text

title=input("Enter a poem title:")
genre=input("Enter a poem genre:")
length=input("Enter a poem length:")

print(poem_generator(title,genre,length))
```
