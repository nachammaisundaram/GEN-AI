```python
system_instruction="""
You are a helpful and knowledgeable tutor.

Your job is to guide the user to find the answer on their own.

* Never give the direct answer, even if the user asks for it.
* Give simple hints, clues, definitions, examples, or questions that help the user reach the answer.
* Make each hint gradually easier and closer to the answer.
* Use the user's previous responses to decide the next hint.
* If the user asks for the direct answer, politely explain that your role is to guide them and continue with a simpler hint.
* Use simple language and avoid complex terminology.
* Once the user reaches the correct answer, confirm it and give a short summary."""

MODEL="gemini-3.5-flash-lite"

chat=client.chats.create(
    model=MODEL,
    config=types.GenerateContentConfig(
        system_instruction=system_instruction,
        temperature=0.7
    ),
    history=[]
)

print(" ____AI Socratic Tutor___")
print("Enter exit or bye to end the conversation")

while True:
  user_input=input("You:")
  if user_input.lower() in['exit','bye']:
    print("Well done! See you in the next challenge!")
    break

  response=chat.send_message(user_input)
  print(f"Bot:{response.text}")

show_history = input("Do you want to see the chat history? (yes/no): ")
if show_history.lower() == "yes":
  print()
  print("___________CHAT HISTORY___________")
  for i, msg in enumerate(chat.get_history()):
    if msg.role=="user":
      role="You"
    else:
      role="Bot"
    print(f"{role}: {msg.parts[0].text[:100]}")
else:
  print("Thank You!")
```
