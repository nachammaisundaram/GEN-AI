```python
#import and install libraries
from google import genai
from google.genai import types
from google.colab import userdata
```

```python
#config
client=genai.Client(api_key=userdata.get('Ragproject'))
```

```python
for i in client.models.list():
  print(i.name)
```

```python
MODEL='gemini-3.5-flash-lite'
response=client.models.generate_content(
    model=MODEL,
    contents="tell me a one-line joke about Tamil Nadu politics",
    config=types.GenerateContentConfig
    (
        system_instruction="you are a helpful assistant",
        temperature=0.5,
        max_output_tokens=200,
        thinking_config=types.ThinkingConfig(thinking_level="low")
    )
)

print(f"Text generated by AI:\n{response.text}")
print()
u=response.usage_metadata
print("prompt_token_count:",u.prompt_token_count)
print("candidate_token_count:",u.candidates_token_count)
print("total_token_count:",u.total_token_count)
```
